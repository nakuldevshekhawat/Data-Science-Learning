 # NumPy Mathematical Operations



 NumPy supports mathematical operations on arrays, including element-wise operations and matrix multiplication.

 ### 1. List Addition vs NumPy Array Addition



 **Purpose:** Understand the difference between Python list operations and NumPy array operations.



 - Python list `+` concatenates two lists.

 - NumPy array `+` performs element-wise addition.



 **Syntax:** `arr1 + arr2`

````python
import numpy as np
````

````python
a = [1, 2, 3, 4]
b = [3, 4, 5, 6]

print(a + b)
# Output: [1, 2, 3, 4, 3, 4, 5, 6]
````

````python
a1 = np.array(a)
b1 = np.array(b)

print(a1 + b1)
# Output: [4 6 8 10]
````

 ### 2. Element-wise Multiplication



 **Purpose:** Multiplies corresponding elements of two arrays.



 - Both arrays must have compatible shapes.

 - Each element is multiplied with the corresponding element.



 **Syntax:** `arr1 * arr2`

````python
print(a1 * b1)
# Output: [3 8 15 24]
````

 ### 3. np.dot()



 **Purpose:** Performs dot product or matrix multiplication depending on the dimensions of the arrays.



 **For 1D arrays:**

 - Multiplies corresponding elements.

 - Adds all the multiplication results.



 **Syntax:** `np.dot(arr1, arr2)`

````python
print(np.dot(a1, b1))
# Output: 50
# (1*3) + (2*4) + (3*5) + (4*6)
````

 ### 4. Matrix Multiplication (2D Arrays)



 **Purpose:** Multiplies two matrices.



 - Number of columns in the first matrix must equal the number of rows in the second matrix.

 - If A has shape (m, n), B must have shape (n, p).

 - The resulting matrix has shape (m, p).



 **Syntax:** `np.dot(A, B)` or `A @ B`

````python
a2 = np.array([[1, 2, 3],
               [2, 3, 4]])

b2 = np.array([[1, 2],
               [2, 3],
               [3, 4]])
````

````python
print(np.dot(a2, b2))

# Output:
# [[14 20]
#  [20 29]]
````

 ### 5. Matrix Multiplication Using Transpose



 **Purpose:** Transpose changes rows into columns and columns into rows.



 - Use `.T` to transpose a 2D array.

 - Transpose can help make matrix dimensions compatible.



 **Syntax:** `arr.T`

````python
a2 = np.array([[1, 2, 3],
               [2, 3, 4]])

b2 = np.array([[1, 2, 3],
               [2, 3, 4]])
````

````python
print("Shape of A:", a2.shape)
print("Shape of B:", b2.shape)
````

````python
# Matrix multiplication using transpose
print(np.dot(a2, b2.T))

# Output:
# [[14 20]
#  [20 29]]
````

 ### 6. Real-World Example: Sales Calculation



 **Scenario:** An Amazon store sells five products in two different states.



 We have:

 - Quantity sold in each state.

 - Price of each product in each state.

 - Total sales revenue for each state.



 **Formula:**



 Total Revenue = Quantity Sold × Product Price (summed across products)

````python
# Apple, Banana, Smartphone, Toy, Furniture

shop_sold_amazon = np.array([
    [100, 200, 20, 30, 5],   # State 1
    [200, 400, 40, 60, 10]   # State 2
])
````

````python
price = np.array([
    [120, 50, 20000, 300, 5000],  # State 1
    [200, 400, 40000, 600, 10000] # State 2
])
````

 **Incorrect Approach:**



 Using matrix multiplication directly creates cross-state calculations as well.

 It does not directly represent the individual state revenue.

````python
sales = np.dot(shop_sold_amazon, price.T)
print(sales)
````

 **Correct Approach:**



 - Multiply corresponding quantities and prices.

 - Use `axis=1` to calculate the total revenue of each state.



 **Syntax:** `np.sum(arr1 * arr2, axis=1)`

````python
correct_sales = np.sum(shop_sold_amazon * price, axis=1)

print("State-wise sales:", correct_sales)
````

````python
total_sales = np.sum(correct_sales)

print("Total sales:", total_sales)

# Output:
# State-wise sales: [ 124000  496000]
# Total sales: 620000
````

 # Aggregation Functions



 **Purpose:** Aggregation functions summarize array data by calculating values such as sum, minimum, maximum, mean and standard deviation.



 - Can operate on the entire array.

 - Can operate along a particular axis.

 - Useful in data analysis and statistical calculations.

 ### 1. Basic Aggregation Functions



 | Function | Purpose |

 |---|---|

 | `np.sum()` | Calculates sum |

 | `np.max()` | Returns maximum value |

 | `np.min()` | Returns minimum value |

 | `np.mean()` | Calculates average |

 | `np.std()` | Calculates standard deviation |

 | `np.median()` | Calculates median |

 | `np.argmax()` | Returns index of maximum value |

 | `np.argmin()` | Returns index of minimum value |

 | `np.cumsum()` | Calculates cumulative sum |

````python
a_2 = np.array([1, 2, 3, 4, 5])
````

````python
print("Sum:", np.sum(a_2))
print("Maximum:", np.max(a_2))
print("Minimum:", np.min(a_2))
print("Mean:", np.mean(a_2))
print("Median:", np.median(a_2))
print("Standard Deviation:", np.std(a_2))
````

 ### 2. Aggregation on 2D Arrays



 **Example:**

````python
price = np.array([
    [120, 50, 20000, 300, 5000],  # State 1
    [200, 400, 40000, 600, 10000] # State 2
])
````

````python
print("Total sum:", np.sum(price))
````

 ### 3. Understanding Axis



 **Purpose:** The `axis` parameter specifies the direction along which an aggregation operation is performed.



 For a 2D array:



 - `axis=0`: Performs operation vertically (down the rows), keeping columns.

 - `axis=1`: Performs operation horizontally (across columns), keeping rows.



 **Syntax:** `np.sum(arr, axis=0)` / `np.sum(arr, axis=1)`

````python
# Sum of each column
print(np.sum(price, axis=0))
````

````python
# Sum of each row
print(np.sum(price, axis=1))
````

 **Visual Understanding:**



 Given:



 ```python

 arr = np.array([

     [1, 2, 3],

     [4, 5, 6]

 ])

 ```



 - `axis=0` → [5, 7, 9]

 - `axis=1` → [6, 15]

 ### 4. argmax() and argmin()



 **Purpose:** Returns the index of the maximum or minimum element.



 - `np.argmax()` returns the index of the maximum element.

 - `np.argmin()` returns the index of the minimum element.

 - These are index functions, not aggregation functions.



 **Syntax:** `np.argmax(arr)` / `np.argmin(arr)`

````python
a3 = np.array([1, 20, 3, 4, 5, 60, 38, 307, 29, 5])

print("Maximum element index:", np.argmax(a3))
print("Minimum element index:", np.argmin(a3))

# Output:
# Maximum element index: 7
# Minimum element index: 0
````

 ### 5. Mean and Standard Deviation with Axis

````python
print("Mean:", np.mean(price))
print("Mean row-wise:", np.mean(price, axis=1))
print("Standard deviation row-wise:", np.std(price, axis=1))
````

 ### 6. Cumulative Sum (cumsum)



 **Purpose:** Calculates the running sum of array elements.



 - Each element represents the sum of all previous elements and itself.

 - Useful for cumulative sales, running totals and financial analysis.



 **Syntax:** `np.cumsum(arr)`

````python
print(np.cumsum([1, 2, 3, 4, 5]))

# Output: [1 3 6 10 15]
````

 ### 7. Median



 **Purpose:** Returns the middle value of sorted data.



 - For an odd number of elements, returns the middle element.

 - For an even number of elements, returns the average of the two middle elements.

 - Can calculate median along a particular axis.

````python
print("Column-wise median:", np.median(price, axis=0))
````

````python
print("Row-wise median:", np.median(price, axis=1))
````

 ### Quick Comparison



 | Function | Purpose |

 |---|---|

 | `np.sum()` | Sum of elements |

 | `np.max()` | Maximum value |

 | `np.min()` | Minimum value |

 | `np.mean()` | Average |

 | `np.std()` | Standard deviation |

 | `np.median()` | Middle value |

 | `np.argmax()` | Index of maximum |

 | `np.argmin()` | Index of minimum |

 | `np.cumsum()` | Running sum |

 # Sorting, Searching and Counting



 NumPy provides functions to sort arrays, find unique elements, retrieve sorted indices and count occurrences.

 ### 1. sort()



 **Purpose:** Sorts array elements in ascending order by default.



 - Returns a sorted copy.

 - Original array remains unchanged.

 - Works with 1D and multidimensional arrays.



 **Syntax:** `np.sort(arr)`

````python
arr = np.array([3, 4, 5, 75, 324, 63, 1, 43, 631, 45])
````

````python
print("Original:", arr)
print("Sorted:", np.sort(arr))
````

 ### 2. argsort()



 **Purpose:** Returns the indices that would sort an array.



 - Returns indices instead of sorted elements.

 - Useful when you want to maintain the relationship between values and their original positions.



 **Syntax:** `np.argsort(arr)`

````python
print("Original indices of sorted elements:", np.argsort(arr))
````

 ### 3. unique()



 **Purpose:** Returns the sorted unique elements of an array.



 - Removes duplicate values.

 - Returns elements in sorted order by default.



 **Syntax:** `np.unique(arr)`

````python
print("Unique elements:", np.unique(arr))
````

 ### 4. bincount()



 **Purpose:** Counts the occurrences of each non-negative integer in an array.



 - Index represents the element.

 - Value at that index represents its frequency.

 - Works with non-negative integer arrays.



 **Syntax:** `np.bincount(arr)`

````python
arr1 = np.array([1, 2, 3, 4, 5, 6, 7, 8,
                 1, 3, 4, 5, 2, 4, 6, 7, 4, 5])
````

````python
print("Count of each element:", np.bincount(arr1))
````

 **Example:**



 If `arr = [1, 2, 2, 3]`



 Output: `[0, 1, 2, 1]`



 - Index 0 → 0 occurrences

 - Index 1 → 1 occurrence

 - Index 2 → 2 occurrences

 - Index 3 → 1 occurrence

 ### 5. Iterating Through Arrays



 **Purpose:** Access array elements one by one using loops.



 - Iterating over a 1D array gives individual elements.

 - Iterating over a 2D array gives individual rows.

 - Use nested loops to access individual elements in a 2D array.

````python
arr1 = np.array([1, 2, 3, 4, 5])

for i in arr1:
    print(i)
````

 **Iteration in 2D Array**

````python
arr_2d = np.array([[1, 2],
                   [3, 4]])

for i in arr_2d:
    print(i)
````

 **Nested Loop for 2D Array**

````python
for row in arr_2d:
    for element in row:
        print(element)
````

 ### 6. Creating 3D Arrays



 **Purpose:** Create multidimensional arrays using `np.arange()` and `reshape()`.



 - `np.arange(n)` generates numbers from 0 to n-1.

 - `reshape()` changes the dimensions without changing the number of elements.

 - `ndim` returns the number of dimensions.



 **Syntax:** `np.arange(start, stop, step)`

````python
arr2 = np.arange(8).reshape(2, 2, 2)

print(arr2)
````

````python
print("Number of dimensions:", arr2.ndim)
print("Shape:", arr2.shape)
````

 # NumPy Character and String Operations



 NumPy provides the `np.char` module for performing vectorized string operations on arrays.



 - Performs string operations on array elements.

 - Supports concatenation, case conversion, replacement, searching and counting.

 - Useful for basic text preprocessing.

 ### 1. add()



 **Purpose:** Concatenates corresponding strings from two arrays.



 - Joins strings element-wise.

 - Both arrays should have compatible shapes.



 **Syntax:** `np.char.add(arr1, arr2)`

````python
a = np.array(['hello', 'world'])
b = np.array(['_user', '_123'])
````

````python
print(np.char.add(a, b))

# Output:
# ['hello_user' 'world_123']
````

 ### 2. multiply()



 **Purpose:** Repeats each string a specified number of times.



 **Syntax:** `np.char.multiply(arr, number)`

````python
print(np.char.multiply(a, 2))

# Output:
# ['hellohello' 'worldworld']
````

 ### 3. String Case Conversion



 | Function | Purpose |

 |---|---|

 | `np.char.upper()` | Converts to uppercase |

 | `np.char.lower()` | Converts to lowercase |

 | `np.char.capitalize()` | Capitalizes the first character |

````python
print("Uppercase:", np.char.upper(a))
print("Lowercase:", np.char.lower(a))
print("Capitalize:", np.char.capitalize(a))
````

 ### 4. replace()



 **Purpose:** Replaces a specified substring with another substring.



 **Syntax:** `np.char.replace(arr, old, new)`

````python
a = np.array(['data_science', 'machine_learning'])

print(np.char.replace(a, '_', '&'))

# Output:
# ['data&science' 'machine&learning']
````

 ### 5. split()



 **Purpose:** Splits strings into smaller parts using a specified separator.



 - Returns an array containing lists of split strings.

 - Default separator is whitespace.



 **Syntax:** `np.char.split(arr, sep='_')`

````python
print(np.char.split(a, sep='_'))

# Output:
# [list(['data', 'science']) list(['machine', 'learning'])]
````

 ### 6. find()



 **Purpose:** Finds the first occurrence of a substring.



 - Returns the starting index of the substring.

 - Returns -1 if the substring is not found.



 **Syntax:** `np.char.find(arr, substring)`

````python
print(np.char.find(a, 'learn'))

# Output:
# [-1 8]
````

 ### 7. count()



 **Purpose:** Counts how many times a substring occurs in each string.



 - Returns the number of occurrences.

 - Case-sensitive operation.



 **Syntax:** `np.char.count(arr, substring)`

````python
print(np.char.count(a, 'a'))
print(np.char.count(a, 'at'))
````

 ### Quick Comparison



 | Function | Purpose |

 |---|---|

 | `np.char.add()` | Concatenates strings |

 | `np.char.multiply()` | Repeats strings |

 | `np.char.upper()` | Uppercase conversion |

 | `np.char.lower()` | Lowercase conversion |

 | `np.char.capitalize()` | Capitalizes first character |

 | `np.char.replace()` | Replaces substring |

 | `np.char.split()` | Splits strings |

 | `np.char.find()` | Finds substring position |

 | `np.char.count()` | Counts substring occurrences |
