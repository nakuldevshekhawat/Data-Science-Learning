```python
# apple,banana,smartphone,toy,furniture
shop_sold_amazon=np.array([
    [100,200,20,30,5],#state 1 inventory
    [200,400,40,60,10]#state 2 inventory
])
```

```python
price=np.array([
    [120,50,20000,300,5000],#state 1 inventory
    [200,400,40000,600,10000]#state 2 inventory
])
```

```python
sales=np.dot(shop_sold_amazon,price.T)
sales#wrong answer
```

```python
diag=np.eye(2)
correct_sales=diag*sales
correct_sales
```

```python
total_sales=np.sum(correct_sales)
```

# Aggregation Function

```python
a_2=np.array([1,2,3,4,5])
```

```python
np.sum(a_2)
```

```python
np.max(a_2)
```

```python
np.min(a_2)
```

```python
price=np.array([
    [120,50,20000,300,5000],#state 1 inventory
    [200,400,40000,600,10000]#state 2 inventory
])
```

```python
np.sum(price)
```

```python
np.sum(price,axis=0)
# axis->0 means sum over rows ,keep columns
```

```python
np.sum(price,axis=1)
# axis->1 means sum over columns ,keep rows
```

```python
##Index of the max element , # not aggregate function
a3=np.array([1,20,3,4,5,60,38,307,29,5])
np.argmax(a3)
```

```python
np.mean(price)
```

```python
np.mean(price,axis=1)
```

```python
np.std(price,axis=1)
```

```python
#cummalative sum
np.cumsum([1,2,3,4,5])
```

```python
np.median(price,axis=0)
```

```python
np.median(price,axis=1)
```
