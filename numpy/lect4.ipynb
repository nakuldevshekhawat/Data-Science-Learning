 ## 1. Additional Array Creation Functions



 ### 1.1 `np.ones()`



 **Purpose:** Creates an array filled with ones.



 - Accepts an integer or tuple to define the shape.

 - Default data type is `float64`.

 - Useful for initializing arrays.



 **Syntax:** `np.ones(shape, dtype=None)`

```python
import numpy as np

# 1D array
a = np.ones(5)
print(a)

# 2D array
b = np.ones((2, 3))
print(b)

# Specify data type
c = np.ones((2, 3), dtype=int)
print(c)
```

 ### 1.2 `np.full()`



 **Purpose:** Creates an array filled with a specified value.



 **Syntax:** `np.full(shape, fill_value, dtype=None)`

```python
a = np.full((2, 3), 7)
print(a)

b = np.full(5, 2.5)
print(b)
```

 ### 1.3 `np.empty()`



 **Purpose:** Creates an array without initializing its entries to a particular value.



 - The values may appear arbitrary because the memory is not initialized.

 - Do not assume the array contains zeros.

 - Assign values before using its contents.



 **Syntax:** `np.empty(shape, dtype=float)`

```python
a = np.empty((2, 3))
print(a)

# Initialize before relying on the values
a[:] = 0
print(a)
```

 ### 1.4 `np.eye()` and `np.identity()`



 **Purpose:** Create identity-like matrices.



 - `np.eye(N, M=None, k=0)` creates a 2D array with ones on a selected diagonal.

 - `np.identity(n)` creates an n × n identity matrix.

 - An identity matrix has ones on the main diagonal and zeros elsewhere.

```python
print(np.eye(3))

# Rectangular matrix
print(np.eye(3, 4))

# Shift the diagonal
print(np.eye(3, k=1))

print(np.identity(3))
```

 ### 1.5 Random Array Creation



 NumPy provides random number generation through `np.random`.



 - `np.random.rand()` generates samples from a uniform distribution over [0, 1).

 - `np.random.randn()` generates samples from the standard normal distribution.

 - `np.random.randint()` generates random integers from a low value (inclusive) to a high value (exclusive).

```python
print(np.random.rand(2, 3))
print(np.random.randn(2, 3))
print(np.random.randint(1, 10, size=(2, 3)))
```

 ## 2. Array Attributes and Data Type Conversion



 ### 2.1 Important Array Attributes



 | Attribute | Purpose |

 |---|---|

 | `arr.shape` | Returns dimensions along each axis |

 | `arr.size` | Returns total number of elements |

 | `arr.ndim` | Returns number of dimensions |

 | `arr.dtype` | Returns element data type |

 | `arr.itemsize` | Returns bytes used by one element |

 | `arr.nbytes` | Returns total bytes used by array elements |



 **Syntax:** `arr.attribute`

```python
arr = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int32)

print("Shape:", arr.shape)
print("Size:", arr.size)
print("Dimensions:", arr.ndim)
print("Data type:", arr.dtype)
print("Item size:", arr.itemsize)
print("Total bytes:", arr.nbytes)
```

 ### 2.2 `astype()`



 **Purpose:** Converts an array to another data type.



 - Returns a new array with the requested data type.

 - The original array is not changed unless the result is assigned back.



 **Syntax:** `arr.astype(new_dtype)`

```python
arr = np.array([1.2, 2.8, 3.5])

int_arr = arr.astype(int)

print("Original:", arr)
print("Converted:", int_arr)
print("New data type:", int_arr.dtype)
```

 ## 3. Combining Arrays



 ### 3.1 `np.concatenate()`



 **Purpose:** Joins arrays along an existing axis.



 - Arrays must have matching dimensions except along the joining axis.

 - `axis=0` joins along rows for 2D arrays.

 - `axis=1` joins along columns for 2D arrays.



 **Syntax:** `np.concatenate((arr1, arr2), axis=0)`

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("Axis 0:")
print(np.concatenate((a, b), axis=0))

print("Axis 1:")
print(np.concatenate((a, b), axis=1))
```

 ### 3.2 `np.vstack()` and `np.hstack()`



 - `np.vstack()` stacks arrays vertically (row-wise).

 - `np.hstack()` stacks arrays horizontally (column-wise for 2D arrays).



 **Syntax:** `np.vstack((arr1, arr2))` / `np.hstack((arr1, arr2))`

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Vertical stack:")
print(np.vstack((a, b)))

print("Horizontal stack:")
print(np.hstack((a, b)))
```

 ### 3.3 `np.stack()`



 **Purpose:** Joins a sequence of arrays along a new axis.



 - Unlike `concatenate()`, it creates a new axis.

 - Input arrays must have the same shape.



 **Syntax:** `np.stack((arr1, arr2), axis=0)`

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.stack((a, b), axis=0))
print(np.stack((a, b), axis=1))
```

 ## 4. Splitting Arrays



 ### 4.1 `np.split()`



 **Purpose:** Splits an array into multiple sub-arrays.



 - If an integer number of sections is provided, the array must be divisible equally.

 - A list of split indices can be used for uneven splits.



 **Syntax:** `np.split(arr, sections, axis=0)`

```python
arr = np.array([1, 2, 3, 4, 5, 6])

print(np.split(arr, 3))
print(np.split(arr, [2, 4]))
```

 ### 4.2 `np.hsplit()` and `np.vsplit()`



 - `np.hsplit()` splits a 2D array horizontally, across columns.

 - `np.vsplit()` splits a 2D array vertically, across rows.

```python
arr = np.arange(16).reshape(4, 4)

print("Horizontal split:")
print(np.hsplit(arr, 2))

print("Vertical split:")
print(np.vsplit(arr, 2))
```

 ## 5. Adding, Inserting and Deleting Elements



 ### 5.1 `np.append()`



 **Purpose:** Appends values to an array and returns a new array.



 - Does not modify the original array.

 - If `axis` is omitted, the input is flattened before appending.

```python
arr = np.array([1, 2, 3])

new_arr = np.append(arr, [4, 5])

print("Original:", arr)
print("New array:", new_arr)
```

 ### 5.2 `np.insert()`



 **Purpose:** Inserts values before the specified index and returns a new array.



 **Syntax:** `np.insert(arr, index, values, axis=None)`

```python
arr = np.array([10, 20, 30, 40])

new_arr = np.insert(arr, 2, 99)

print(new_arr)
```

 ### 5.3 `np.delete()`



 **Purpose:** Returns a new array with selected elements or sub-arrays removed.



 **Syntax:** `np.delete(arr, index, axis=None)`

```python
arr = np.array([10, 20, 30, 40])

new_arr = np.delete(arr, 1)

print("Original:", arr)
print("After deletion:", new_arr)
```

 ## 6. Expanding and Changing Dimensions



 ### 6.1 `np.expand_dims()`



 **Purpose:** Adds a new dimension of size 1 at the specified axis.



 **Syntax:** `np.expand_dims(arr, axis)`

```python
arr = np.array([1, 2, 3])

row = np.expand_dims(arr, axis=0)
column = np.expand_dims(arr, axis=1)

print("Original shape:", arr.shape)
print("Row shape:", row.shape)
print("Column shape:", column.shape)
```

 ### 6.2 `np.squeeze()`



 **Purpose:** Removes dimensions of size 1.



 - Can remove all size-1 dimensions.

 - `axis` can specify which dimension to remove.

 - The selected axis must have size 1.



 **Syntax:** `np.squeeze(arr, axis=None)`

```python
arr = np.array([[[1], [2], [3]]])

print("Original shape:", arr.shape)

squeezed = np.squeeze(arr)
print("After squeeze:", squeezed.shape)
```

 ### 6.3 `np.transpose()` and `np.swapaxes()`



 - `np.transpose()` permutes the axes of an array.

 - For a 2D array, `.T` is a convenient transpose.

 - `np.swapaxes()` exchanges two specified axes.



 **Syntax:** `np.transpose(arr, axes=None)` / `np.swapaxes(arr, axis1, axis2)`

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])

print("Transpose:")
print(np.transpose(arr))
print(arr.T)

arr3d = np.arange(24).reshape(2, 3, 4)
swapped = np.swapaxes(arr3d, 0, 2)

print("Original 3D shape:", arr3d.shape)
print("Swapped shape:", swapped.shape)
```

 ## 7. Mathematical and Statistical Functions



 ### 7.1 Element-wise Mathematical Functions



 These functions operate on each element of an array.



 | Function | Purpose |

 |---|---|

 | `np.add()` | Addition |

 | `np.subtract()` | Subtraction |

 | `np.multiply()` | Multiplication |

 | `np.divide()` | Division |

 | `np.sqrt()` | Square root |

 | `np.power()` | Power |

 | `np.exp()` | Exponential |

 | `np.log()` | Natural logarithm |

 | `np.abs()` | Absolute value |

 | `np.round()` | Rounding |

 | `np.floor()` | Round down |

 | `np.ceil()` | Round up |

```python
a = np.array([1, 4, 9, 16])
b = np.array([1, 2, 3, 4])

print("Add:", np.add(a, b))
print("Subtract:", np.subtract(a, b))
print("Multiply:", np.multiply(a, b))
print("Divide:", np.divide(a, b))

print("Square root:", np.sqrt(a))
print("Power:", np.power(b, 2))
print("Absolute:", np.abs([-2, -1, 0, 1, 2]))

values = np.array([1.2, 2.5, 3.8])
print("Round:", np.round(values))
print("Floor:", np.floor(values))
print("Ceil:", np.ceil(values))
```

 ### 7.2 Variance and Percentile



 - `np.var()` measures the average squared deviation from the mean.

 - `np.percentile()` returns the value below which a given percentage of observations falls.

```python
arr = np.array([10, 20, 30, 40, 50])

print("Variance:", np.var(arr))
print("25th percentile:", np.percentile(arr, 25))
print("50th percentile:", np.percentile(arr, 50))
print("90th percentile:", np.percentile(arr, 90))
```

 ### 7.3 `np.any()` and `np.all()`



 - `np.any()` returns `True` if at least one element satisfies the condition.

 - `np.all()` returns `True` only if every element satisfies the condition.

 - Both can use the `axis` parameter.

```python
arr = np.array([2, 4, 6, 8])

print("Any element greater than 5:", np.any(arr > 5))
print("All elements greater than 0:", np.all(arr > 0))

matrix = np.array([[1, 2], [3, 4]])
print("Any by column:", np.any(matrix > 2, axis=0))
print("All by row:", np.all(matrix > 0, axis=1))
```

 ## 8. Conditional Operations and Filtering



 ### 8.1 `np.where()`



 **Purpose:** Returns indices where a condition is true, or selects values depending on a condition.



 **Two common forms:**

 - `np.where(condition)` returns indices of true elements.

 - `np.where(condition, x, y)` selects from `x` where true and `y` where false.

```python
arr = np.array([10, 25, 30, 5, 40])

print("Indices:", np.where(arr > 20))
print("Selected values:", np.where(arr > 20, arr, 0))
```

 ### 8.2 `np.clip()`



 **Purpose:** Limits values to a specified minimum and maximum range.



 - Values below the minimum become the minimum.

 - Values above the maximum become the maximum.

```python
arr = np.array([5, 15, 25, 35, 45])

print(np.clip(arr, 10, 30))
```

 ### 8.3 `np.corrcoef()`



 **Purpose:** Calculates the correlation coefficient matrix.



 - Correlation values range from -1 to +1.

 - +1 indicates perfect positive linear correlation.

 - -1 indicates perfect negative linear correlation.

 - 0 indicates no linear correlation.

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

print(np.corrcoef(x, y))
```

 ## 9. Broadcasting



 **Purpose:** Broadcasting allows NumPy to perform operations on arrays with different shapes when their dimensions are compatible.



 ### Broadcasting Rules



 Compare shapes from right to left. Two dimensions are compatible when:



 1. They are equal, or

 2. One of them is 1.



 If neither condition is met, NumPy raises a broadcasting error.



 **Examples:**

 - `(3, 1)` and `(1, 4)` → compatible; result shape `(3, 4)`.

 - `(2, 3)` and `(3,)` → compatible; result shape `(2, 3)`.

 - `(2, 3)` and `(2,)` → not compatible.

```python
# Scalar broadcasting
arr = np.array([1, 2, 3])
print(arr + 10)

# 2D array and 1D array
a = np.array([[1, 2, 3],
              [4, 5, 6]])

b = np.array([10, 20, 30])

print(a + b)

# Column vector and row vector
x = np.array([[1], [2], [3]])   # Shape (3, 1)
y = np.array([10, 20, 30, 40])  # Shape (4,)

result = x + y
print(result)
print("Result shape:", result.shape)
```

 ## 10. Random Number Generation



 ### 10.1 Reproducible Random Numbers



 A random seed makes a sequence of pseudo-random values reproducible.



 **Legacy syntax:** `np.random.seed(seed)`



 For new code, NumPy recommends creating a generator with `np.random.default_rng()`.

```python
# Legacy random API
np.random.seed(42)
print(np.random.randint(1, 10, size=5))

# Recommended Generator API
rng = np.random.default_rng(42)

print(rng.integers(1, 10, size=5))
print(rng.random(5))
print(rng.normal(size=5))
```

 ### 10.2 Common Generator Methods



 | Method | Purpose |

 |---|---|

 | `rng.random()` | Uniform random floats in [0, 1) |

 | `rng.integers()` | Random integers |

 | `rng.normal()` | Normal distribution samples |

 | `rng.choice()` | Random selection from given values |

 | `rng.permutation()` | Returns a random permutation |

 | `rng.shuffle()` | Shuffles an array in place |

```python
rng = np.random.default_rng(10)

arr = np.array([10, 20, 30, 40, 50])

print("Random choice:", rng.choice(arr, size=3))
print("Permutation:", rng.permutation(arr))

shuffled = arr.copy()
rng.shuffle(shuffled)
print("Shuffled copy:", shuffled)
print("Original:", arr)
```

 ## 11. Handling Missing and Invalid Values



 ### 11.1 NaN and Infinite Values



 - `np.nan` represents Not a Number.

 - `np.inf` represents positive infinity.

 - `-np.inf` represents negative infinity.



 ### 11.2 Checking Values



 | Function | Purpose |

 |---|---|

 | `np.isnan()` | Checks for NaN |

 | `np.isinf()` | Checks for infinity |

 | `np.isfinite()` | Checks for finite values |

```python
arr = np.array([1.0, np.nan, np.inf, -np.inf, 5.0])

print("NaN:", np.isnan(arr))
print("Infinite:", np.isinf(arr))
print("Finite:", np.isfinite(arr))
```

 ### 11.3 NaN-aware Aggregation



 Regular aggregation functions such as `np.mean()` can return NaN when the input contains NaN.



 NaN-aware alternatives include:



 - `np.nanmean()`

 - `np.nanmedian()`

 - `np.nansum()`

 - `np.nanmax()`

 - `np.nanmin()`



 These functions generally ignore NaN values. Check the documentation for special cases, such as all-NaN slices.

```python
arr = np.array([10.0, 20.0, np.nan, 40.0])

print("Regular mean:", np.mean(arr))
print("NaN-aware mean:", np.nanmean(arr))
print("NaN-aware median:", np.nanmedian(arr))
print("NaN-aware sum:", np.nansum(arr))
```

 ## 12. Practical Example: Basic Data Analysis



 Use multiple NumPy operations together to analyze a small set of marks.



 Tasks:

 1. Store student marks in an array.

 2. Calculate average marks.

 3. Find the highest and lowest marks.

 4. Find students scoring at least 60.

 5. Replace marks below 40 with 40 using `np.clip()`.

```python
marks = np.array([35, 78, 92, 56, 41, 88, 29])

print("Marks:", marks)
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))

print("Students scoring at least 60:", marks[marks >= 60])
print("Marks clipped to minimum 40:", np.clip(marks, 40, 100))
```
