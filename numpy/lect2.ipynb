# Array resizing and reshaping


### 1. reshape()

**Purpose:** Changes the shape of an array without changing its elements.

- Total number of elements must remain the same.
- Returns a view if possible; otherwise, a copy.
- Original array shape remains unchanged.
- Use `-1` to automatically calculate one dimension.

**Syntax:** `arr.reshape(rows, columns)`

```python

a = np.array([[1, 2, 3], [4, 5, 6]])

# Change 2x3 into 3x2
b = a.reshape(3, 2)
print(b)
print("Original shape:", a.shape)
print("New shape:", b.shape)


# Using -1 (automatic dimension calculation)
print(a.reshape(3, -1))
print(a.reshape(-1))


# Check whether memory is shared
print(np.shares_memory(a, b))
```

- Returns a view whenever possible; otherwise, a copy.
- Memory-efficient compared to `flatten()`.
- Changes reflect in the original if memory is shared.
- Does not change the original array's shape.

**Syntax:** `arr.ravel()`

### 3. flatten()

**Purpose:** Converts a multidimensional array into a 1D array.

- Always returns a copy of the original array.
- Changes in the flattened array do not affect the original.
- Uses additional memory compared to `ravel()`.

**Syntax:** `arr.flatten()`

### 4. resize()

**Purpose:** Changes the shape and total number of elements in an array.

**Two methods:**

- `np.resize(arr, shape)`
  - Returns a new array.
  - Repeats elements if the new size is larger.
  - Truncates elements if the new size is smaller.
  - Original array remains unchanged.

- `arr.resize(shape)`
  - Modifies the original array (in-place).
  - Returns `None`.
  - Fills additional elements with zeros.
  - May require the array to own its data and have no other references.

**Syntax:** `np.resize(arr, shape)` / `arr.resize(shape)`

### 5. squeeze()

**Purpose:** Removes dimensions having size 1 from an array.

- Returns a view whenever possible.
- Original array shape remains unchanged.
- Can remove a specific dimension using the `axis` parameter.
- Useful in image processing and deep learning.

**Syntax:** `np.squeeze(arr)` / `np.squeeze(arr, axis=0)`

```python
a1=np.array([[[1],[2],[3]]])
print(a1.shape)
```

```python
b=np.squeeze(a1)
```

```python
print(b.shape)
```

### Quick Comparison

| Function | Purpose | Original Modified? | In-Place? | Returns |
|---|---|---|---|---|
| `reshape()` | Changes shape | No | No | View/copy |
| `np.resize()` | Changes shape and size | No | No | New array |
| `arr.resize()` | Changes shape and size | Yes | Yes | None |
| `ravel()` | Flattens to 1D | No | No | View/copy |
| `flatten()` | Flattens to 1D | No | No | Copy |
| `squeeze()` | Removes size-1 axes | No | No | View (usually) |

## Indexing and Slicing

**Indexing:** Accessing individual elements of an array using their position (index).

**Slicing:** Extracting a portion of an array using a range of indices.

- NumPy uses zero-based indexing.
- Negative indexing accesses elements from the end.
- Supports 1D, 2D and multidimensional arrays.
- Slicing generally returns a view, not a copy.

**Syntax:**
- Indexing: `arr[index]`
- Slicing: `arr[start:stop:step]`

### 1. Basic Indexing (1D)

- Access individual elements using their index.
- Indexing starts from 0.
- Negative indexing starts from -1 (last element).

```python
arr=np.array([1,2,3,4,5,6])
```

```python
print(arr[0])
print("last")

print(arr[-1])
```

```python
# slicing
print(arr[2:5])
```

```python
# reversing
arr[::-1]
```

# 4. Fancy Indexing

**Purpose:** Access multiple specific elements using a list or array of indices.

- Can access non-consecutive elements.
- Returns a copy, not a view.
- Original array remains unchanged when modifying the result.

**Syntax:** `arr[[index1, index2, index3]]`

```python

arr = np.array([10, 20, 30, 40, 50, 60])

print(arr[[0, 2, 4]])   # [10 30 50]
print(arr[[1, 3, 5]])   # [20 40 60]

# Fancy indexing in 2D
b = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

print(b[[0, 2], [1, 2]])  # [20 90]
```

### 5. View vs Copy

**View:**
- Shares memory with the original array.
- Changes may reflect in the original array.
- Basic slicing generally returns a view.

**Copy:**
- Has independent memory.
- Changes do not affect the original array.
- Fancy indexing and Boolean indexing return copies.

**Check:** `np.shares_memory(arr, result)`

```python
# %%
arr = np.array([10, 20, 30, 40, 50])

# Basic slicing (View)
a = arr[1:4]

# Fancy indexing (Copy)
b = arr[[1, 2, 3]]

print("Slicing shares memory:", np.shares_memory(arr, a))
print("Fancy indexing shares memory:", np.shares_memory(arr, b))

# Modify view
a[0] = 999

# Modify copy
b[0] = 888

print("Original:", arr)
print("View:", a)
print("Copy:", b)
```

### Quick Comparison

| Operation | Purpose | Returns View/Copy? | Modifies Original? |
|---|---|---|---|
| `arr[i]` | Access single element | Scalar | No |
| `arr[i:j]` | Basic slicing | View (generally) | No |
| `arr[row, col]` | Access 2D element | Scalar | No |
| `arr[[i,j]]` | Fancy indexing | Copy | No |
| `arr[arr > 10]` | Boolean indexing | Copy | No |
| `arr[::-1]` | Reverse array | View (generally) | No |
| `arr[i] = value` | Modify element | N/A | Yes |
