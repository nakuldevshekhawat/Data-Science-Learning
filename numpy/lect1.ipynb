 # NumPy Array Creation



 NumPy provides different methods to create arrays efficiently.



 - `np.array()` creates arrays from Python lists or tuples.

 - `np.arange()` generates arrays with a specified interval.

 - `np.linspace()` generates evenly spaced values.

 - `np.zeros()` creates arrays filled with zeros.

```python
import numpy as np
```

 ### 1. np.array()



 **Purpose:** Converts Python lists or tuples into NumPy arrays.



 - Creates a NumPy ndarray.

 - Supports one-dimensional and multidimensional arrays.

 - Automatically determines the data type if `dtype` is not specified.

 - Can explicitly define the data type using `dtype`.



 **Syntax:** `np.array(object, dtype=None)`

```python
array_n1 = np.array([1, 2, 3])

print(array_n1)
print(type(array_n1))
print(type(array_n1[0]))
```

 **Specifying Data Type (dtype)**



 - `dtype` specifies the data type of array elements.

 - NumPy uses a common data type for all elements in an array.

 - If different data types are provided, NumPy performs type promotion.



 **Syntax:** `np.array(list, dtype=data_type)`

```python
array_n2 = np.array([1, 2, "3"], dtype=np.float16)

print(array_n2)
print(type(array_n2))
print(type(array_n2[0]))
```

 **Common NumPy Data Types**



 | Data Type | Description |

 |---|---|

 | `np.int32` | 32-bit integer |

 | `np.int64` | 64-bit integer |

 | `np.float16` | 16-bit floating point |

 | `np.float32` | 32-bit floating point |

 | `np.float64` | 64-bit floating point |

 | `np.bool_` | Boolean |

 | `np.str_` | String |

 ### 2. np.arange()



 **Purpose:** Generates an array containing evenly spaced values within a given interval.



 - Similar to Python's `range()` function.

 - The stop value is excluded.

 - Supports positive and negative step values.

 - Returns a NumPy array.



 **Syntax:** `np.arange(start, stop, step)`



 **Parameters:**

 - `start`: Starting value (default 0).

 - `stop`: Ending value (excluded).

 - `step`: Difference between consecutive values (default 1).

```python
print(np.arange(5))

# Output: [0 1 2 3 4]
```

```python
print(np.arange(1, 10, 2))

# Output: [1 3 5 7 9]
```

```python
arange = np.arange(1, 10, 2)

print(arange)
print(type(arange))
```

 ### 3. np.linspace()



 **Purpose:** Generates a specified number of evenly spaced values between a start and stop value.



 - Includes the stop value by default.

 - Number of elements is controlled by `num`.

 - Useful when equal intervals are required.

 - Returns a NumPy array.



 **Syntax:** `np.linspace(start, stop, num=50, endpoint=True)`



 **Parameters:**

 - `start`: Starting value.

 - `stop`: Ending value.

 - `num`: Number of evenly spaced values (default 50).

 - `endpoint`: Whether to include the stop value (default True).

```python
np_lin = np.linspace(0, 1, 10)

print(np_lin)
print(type(np_lin))
```

```python
# Iterating through linspace values
for i in np.linspace(1, 10, 10):
    print(i)
```

 **Difference Between arange() and linspace()**



 | Feature | np.arange() | np.linspace() |

 |---|---|---|

 | Main parameter | Step size | Number of values |

 | Stop value | Excluded | Included by default |

 | Output | NumPy array | NumPy array |

 | Best use | Fixed intervals | Fixed number of values |



 **Example:**



 - `np.arange(0, 10, 2)` → `[0, 2, 4, 6, 8]`

 - `np.linspace(0, 10, 5)` → `[0, 2.5, 5, 7.5, 10]`

 ### 4. np.zeros()



 **Purpose:** Creates a NumPy array filled with zeros.



 - Accepts an integer or tuple to define the shape.

 - Default data type is `float64`.

 - Useful for initializing arrays before storing data.



 **Syntax:** `np.zeros(shape, dtype=float)`

```python
npz1 = np.zeros(5)

print(npz1)

# Output: [0. 0. 0. 0. 0.]
```

 ### 5. ndim Attribute



 **Purpose:** Returns the number of dimensions (axes) of a NumPy array.



 - 1D array → 1 dimension.

 - 2D array → 2 dimensions.

 - 3D array → 3 dimensions.



 **Syntax:** `arr.ndim`

```python
print("Dimensions:", npz1.ndim)
```

 ### 6. Creating 2D Arrays Using np.zeros()



 **Purpose:** Creates a two-dimensional array filled with zeros.



 - Shape is specified using a tuple `(rows, columns)`.

 - Every element is initialized to zero.

```python
npz2 = np.zeros((5, 5))

print(npz2)
print("Dimensions:", npz2.ndim)
```

 ### 7. Creating 3D Arrays Using np.zeros()



 **Purpose:** Creates a three-dimensional array filled with zeros.



 - Shape is specified using `(depth, rows, columns)`.

 - Useful in image processing and multidimensional data storage.

```python
npz3 = np.zeros((5, 5, 5))

print(npz3)
print("Dimensions:", npz3.ndim)
```

 ### Quick Comparison: Array Creation Functions



 | Function | Purpose | Example |

 |---|---|---|

 | `np.array()` | Creates array from existing data | `np.array([1,2,3])` |

 | `np.arange()` | Creates values using step size | `np.arange(0,10,2)` |

 | `np.linspace()` | Creates a fixed number of evenly spaced values | `np.linspace(0,10,5)` |

 | `np.zeros()` | Creates an array filled with zeros | `np.zeros((2,3))` |

 | `arr.ndim` | Returns number of dimensions | `arr.ndim` |

 # Key Takeaways



 1. Use `np.array()` to convert Python lists into NumPy arrays.

 2. Use `dtype` to control the data type.

 3. Use `np.arange()` when you know the step size.

 4. Use `np.linspace()` when you know the number of values required.

 5. Use `np.zeros()` to initialize arrays with zeros.

 6. Use `ndim` to check the number of dimensions.
